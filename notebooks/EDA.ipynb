```python
# Homepage - признак с наибольшим количество пропусков 

print('Какие уникальные значения есть:', movies['homepage'].unique())
print(f'\n\nКоличество уникальных значений из общего количество не null: {(movies["homepage"].nunique() / movies["homepage"].notnull().sum()):.2f}')
```

**Homepage** = ссылка на страничку в интернете с инфой об этом фильме по факту -> признак довольно не информативный сам по себе 
Ко всему прочему, почти каждое из значений - уникальное, а значит заполнять как-то средним не получиться. 

Я считаю выходом **вообще убрать этот признак, так как он не несет никакой информации для рекомендации фильмов, и заполнение пропусков просто не оправдает своих ожиданий**. 

```python
# Overview - 3 пропуска 

print(movies['overview'].head(1))
print(f'\n\nКоличество уникальных значений из общего количество не null: {(movies["overview"].nunique() / movies["overview"].notnull().sum()):.2f}')
```

**Overview - это описание фильма** -> **важный признак**

```python
# tagline (слоган) - 844 пропуска 
print(f'Количество уникальных значений из общего количество не null: {(movies["tagline"].nunique() / movies["tagline"].notnull().sum()):.2f}')
```

Все taglines, конечно, уникальные и просто так не сгенерить на рандом. Но и заполнять долго, потому что пропусков 844. 
**Принято решение: не обрабатывать эти пропуски, а при обучении обучить две модели: одну с tagline, другую без и проверить разницу.**
Tagline потенциально полезный признак, но 
1. 844 пропуска - это около 17,6% датасета.
2. Слоганы часто содержат рекламные фразы, которые не обязательно отражают содержание фильма.
3. Они могут добавлять шум в сравнение. Например, слова вроде love, fear, life могут встречаться в слоганах совершенно разных фильмов.
